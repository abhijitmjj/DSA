```python
from collections import deque
import itertools
from typing import Generator
```

```python
seq = list(range(1, 11))
iter_seq = iter(seq)
```

```python
[x for x in itertools.islice(iter_seq, 3 - 1)]
```

```python
for elem in iter_seq:
    print(elem)
```

```python
def moving_avg(seq: list[int], n: int) -> Generator[float]:
    iter_seq = iter(seq)
    deq = deque(itertools.islice(seq, n - 1))
    deq.appendleft(0)
    s = sum(deq)
    for elem in iter_seq:
        s += elem - deq.popleft()
        deq.append(elem)
        yield s / n
```

```python
for elem in moving_avg(seq, 3):
    print(elem)
```

```python
def roundrobin(*iterables):
    "roundrobin('ABC', 'D','EF')"
    iterators = deque(map(iter, iterables))
    while iterators:
        try:
            while True:
                yield next(iterators[0])
                iterators.rotate(-1)
        except StopIteration:
            iterators.popleft()
```

```python
for elem in roundrobin("ABC", "D", "EF", ""):
    print(
        elem,
        end=" ",
    )
```

```python
print(
    list(roundrobin("ABC", "DE", "FGHI"))
)  # Should output: ['A', 'D', 'F', 'B', 'E', 'G', 'C', 'H', 'I']
```
