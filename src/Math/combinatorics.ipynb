```python
from __future__ import annotations
import itertools
import math
```

```python
def Combinations(n: int, r: int) -> int:
    return math.comb(n, r)


def Permutations(n: int, r: int) -> int:
    return math.perm(n, r)
```

```python
for x in itertools.combinations([1, 2, 3, 4, 5, 6], 3):
    # exclude 1,2 =>  ensure 1 and 2 together are not in the combination
    if 1 in x and 2 in x:
        continue

    print(x)
```

```python
# number of ways of putting n distinct objects into r containers
for elem in itertools.product(itertools.repeat([1, 2, 3, 4, 5, 6]), repeat=3):
    print(elem)
```
