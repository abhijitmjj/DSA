Okay, let's break down how to reverse a singly linked list using the two-pointers technique. This is a classic interview problem.

**I. Introduction (The Apex Answer)**

Reversing a singly linked list using two (or often three, for clarity) pointers is an efficient in-place algorithm that achieves O(N) time complexity and O(1) space complexity. The core idea is to iterate through the list, changing the `next` pointer of each node to point to its previous node. Two main pointers, `previous` and `current`, track the progress, while a temporary pointer (`next_node`) is often used to store the link to the rest of the list before a node's `next` pointer is redirected.

**II. Core Intuition and Analogy**

Imagine you have a chain of paper clips linked together, representing the linked list. Each clip points to the next. To reverse it:

1.  You pick up the first clip (`current`).
2.  You note which clip it was pointing to (`next_node`) so you don't lose the rest of the chain.
3.  You make this first clip point *backwards* (to `previous`, which is initially nothing or `null`).
4.  Now, this first clip becomes the "new tail" or the `previous` node for the *next* clip you're about to process.
5.  You move to the clip you noted in step 2 (`next_node` becomes the new `current`).
6.  Repeat until you've processed all clips. The last `previous` node you were holding will be the new head of the reversed chain.

**III. Pointer Roles and Movement**

We typically use three conceptual pointers (though sometimes `next_node` is handled implicitly within the loop):

1.  **`previous` (or `prev`):**
    *   **Initial State:** `None` (or `null`), because the new head of the reversed list will initially point to nothing (it will become the new tail of the reversed segment).
    *   **Role:** Tracks the node that the `current` node should point to *after* its `next` pointer is reversed. It always points to the head of the already reversed portion of the list.
    *   **Update:** After `current.next` is updated, `previous` becomes `current`.

2.  **`current` (or `curr`):**
    *   **Initial State:** `head` (the head of the original linked list).
    *   **Role:** The node whose `next` pointer is currently being processed and reversed.
    *   **Update:** After its `next` pointer is reversed and `previous` is updated, `current` moves to `next_node`.

3.  **`next_node` (or `temp_next`):**
    *   **Initial State:** Not strictly initialized before the loop, but set within each iteration.
    *   **Role:** Temporarily stores `current.next` *before* `current.next` is changed. This is crucial because once `current.next` is set to `previous`, we would lose the reference to the rest of the original list if we hadn't saved it.
    *   **Update:** Set to `current.next` at the beginning of each iteration.

**IV. Algorithm Steps (Derivation)**

Let's assume we have a `ListNode` class:
```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

1.  **Initialization:**
    *   `previous = None`
    *   `current = head` (the head of the list to be reversed)

2.  **Iteration:** Loop while `current` is not `None` (i.e., while there are still nodes to process):
    *   **a. Store Next:** `next_node = current.next` (Save the next node before we break the link).
    *   **b. Reverse Pointer:** `current.next = previous` (Make `current` node point to the `previous` node).
    *   **c. Move Pointers Forward:**
        *   `previous = current` (The `current` node is now processed and becomes the new `previous` for the next iteration).
        *   `current = next_node` (Move to the next node in the original list that we saved).

3.  **Termination and New Head:** When the loop finishes, `current` will be `None` (it has moved past the end of the original list). The `previous` pointer will be pointing to the last node processed, which is the new head of the reversed list.
    *   Return `previous`.

**V. Python Implementation**

```python
from typing import Optional # For type hinting

class ListNode:
    def __init__(self, val: int = 0, next: 'Optional[ListNode]' = None):
        self.val = val
        self.next = next

    def __repr__(self) -> str: # Helper for printing
        return f"ListNode({self.val}, next={self.next.val if self.next else None})"

def print_list(head: Optional[ListNode]): # Helper for printing
    nodes = []
    curr = head
    while curr:
        nodes.append(str(curr.val))
        curr = curr.next
    print(" -> ".join(nodes) if nodes else "None")

def reverse_linked_list(head: Optional[ListNode]) -> Optional[ListNode]:
    """
    Reverses a singly linked list using two main pointers (and one temporary).

    Args:
        head: The head of the linked list to be reversed.

    Returns:
        The new head of the reversed linked list.
    """
    previous_node: Optional[ListNode] = None
    current_node: Optional[ListNode] = head
    
    while current_node is not None:
        # 1. Store the next node before overwriting current_node.next
        next_node_temp: Optional[ListNode] = current_node.next 
        
        # 2. Reverse the current node's pointer
        current_node.next = previous_node
        
        # 3. Move pointers one position ahead for the next iteration
        previous_node = current_node   # previous_node moves to where current_node was
        current_node = next_node_temp  # current_node moves to where next_node_temp was
        
    # After the loop, current_node is None, and previous_node is the new head
    return previous_node

# --- Example Usage ---
# Create a linked list: 1 -> 2 -> 3 -> 4 -> None
node4 = ListNode(4)
node3 = ListNode(3, node4)
node2 = ListNode(2, node3)
node1 = ListNode(1, node2)
head_original = node1

print("Original list:")
print_list(head_original)

reversed_head = reverse_linked_list(head_original)

print("\nReversed list:")
print_list(reversed_head) # Expected: 4 -> 3 -> 2 -> 1 -> None

# Test with an empty list
print("\nOriginal list (empty):")
print_list(None)
reversed_head_empty = reverse_linked_list(None)
print("Reversed list (empty):")
print_list(reversed_head_empty) # Expected: None

# Test with a single node list
print("\nOriginal list (single node):")
single_node_head = ListNode(10)
print_list(single_node_head)
reversed_head_single = reverse_linked_list(single_node_head)
print("Reversed list (single node):")
print_list(reversed_head_single) # Expected: 10 -> None
```

**VII. Complexity Analysis**

*   **Time Complexity: O(N)**
    *   The `while` loop iterates through each node of the linked list exactly once. Each operation inside the loop (pointer assignments) takes constant time.
*   **Space Complexity: O(1)**
    *   We only use a few extra pointer variables (`previous_node`, `current_node`, `next_node_temp`) regardless of the size of the list. No additional data structures that scale with N are used. This makes it an in-place reversal.

**VIII. Conclusion**

Reversing a linked list with two (or three for clarity) pointers is a fundamental list manipulation technique. It showcases how local pointer re-wirings, when done systematically, can achieve a global transformation of the data structure efficiently. The key is to carefully manage the order of pointer updates to avoid losing track of nodes, especially by saving `current.next` before overwriting it. This pattern of iterating and re-linking is applicable to many other linked list problems.

````python
from dataclasses import dataclass
from typing import List, Optional
````

````python
@dataclass
class ListNode:
    val: int = -1
    next: "ListNode | None" = None
````

````python
def construct_list_node(values: List[int]) -> Optional[ListNode]:
    if not values:
        return None
    head = ListNode(values[0])
    current = head
    for value in values[1:]:
        current.next = ListNode(value)
        current = current.next
    return head
````

````python
construct_list_node([1, 2, 3, 4, 5])


def print_list_node(node: Optional[ListNode]) -> None:
    current = node
    while current:
        print(current.val, end=" -> ")
        current = current.next
    print("None")


print_list_node(construct_list_node([1, 2, 3, 4, 5]))
````
