```python
import jax
import jax.numpy as jnp

def cumsum_step(carry_sum: jnp.ndarray, x_i: jnp.ndarray) -> tuple[jnp.ndarray, jnp.ndarray]:
    new_sum = carry_sum + x_i
    return new_sum, new_sum # new_carry is new_sum, y_output is also new_sum

data = jnp.array([1.0, 2.0, 3.0, 4.0])
initial_sum = 0.0
final_sum, cumulative_sums = jax.lax.scan(cumsum_step, initial_sum, data)

print(f"Final sum: {final_sum}") # Output: Final sum: 10.0
print(f"Cumulative sums: {cumulative_sums}") # Output: Cumulative sums: [ 1.  3.  6. 10.]
```
